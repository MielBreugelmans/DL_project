# Compare models (paired bootstrap)
Run the bootstrap cell at the bottom of each model notebook first. Positive difference = model B is better. If 0 is not inside the 95% CI, the difference is significant.

```python
import numpy as np
from sklearn.metrics import average_precision_score
from bootstrap import compare_auprc

a = np.load("results/baseline.npz")
b = np.load("results/mednet.npz")
assert (a["patient_ids"] == b["patient_ids"]).all(), "models were not tested on the same test set"
```

```python
auprc_a = average_precision_score(a["y_true"], a["y_prob"])
auprc_b = average_precision_score(b["y_true"], b["y_prob"])
lo, hi, std = compare_auprc(a["y_true"], a["y_prob"], b["y_prob"], a["patient_ids"])

print(f"Baseline AUPRC: {auprc_a:.4f}")
print(f"MedNet AUPRC:   {auprc_b:.4f}")
print(f"Difference: {auprc_b - auprc_a:+.4f} ± {std:.4f}  95% CI [{lo:+.4f}, {hi:+.4f}]")
```
